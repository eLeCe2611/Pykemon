```python
#Importamos la biblioteca pandas bajo el alias 'pd'
import pandas as pd
```

```python
#Ej1: Cargamos en un dataframe panda todos los datos del csv Pokemon
#Usamon la funcion read_csv
pokemon = pd.read_csv("15_pokemon.csv")
#Mostramos los datos almacenados
pokemon
```

```python
#Ej2: Realizamos un filtrado de los datos según los valores de dos de sus columnas
#En este caso mostraremos todos los pokemon legendarios de una generacion concreta, siendo esta elegida por el ususario
gen = int(input('Introduce generacion de la que deseas ver sus legendarios: '))
#Creamos nuevo dataframe donde solo se almacenen los pokemon legendarios de dicha generacion
legplan = pokemon[(pokemon['Generation'] == gen) & (pokemon['Legendary'])]
#Finalmente mostramos el nuevo dataframe
legplan[['Name', 'Generation', 'Legendary']]
```

```python
#Ej3p1: Construimos un nuevo dataframe identico al original, pero al que añadiremos nuevas columna numericas que dependan de las ya existentes
newPokemon = pokemon.copy()
#En este caso añadiremos dos columnas nuevas, que representaran el total de ataque y el total de defensa
#Para ello se sumaran las columna 'Attack' y 'Sp. Atk' para el total de ataque
newPokemon['totAtaque']=pokemon['Attack']+pokemon['Sp. Atk']
#Y 'Defense' y 'Sp. Def' para el total de defensa
newPokemon['totDefensa']=pokemon['Defense']+pokemon['Sp. Def']
#Ademas añadimos una tercera columna que nos indica si el pokemon es mas atacante o defensivo segun el valor de las dos nuevas columnas
#Para ello usamos la funcion apply() que efectuara fila por fila la funcion lambda, la cual se encarga de comprobar si es mayor la columna totAtaque o totDefensa
#axis=1 indica al metodo apply que la funcion debe efectuarse a cada fila individualmente
newPokemon['modoJuego'] = newPokemon.apply(lambda row: 'Atacante' if row['totAtaque'] >= row['totDefensa'] else 'Defensivo', axis=1)
#Finalmente mostramos el nuevo dataframe
newPokemon
```

```python
#Ej3p2: Finalmente agrupamos todos los datos segun un dato relevante
#En este caso creamos un nuevo dataframe donde agrupamos los pokemon segun su tipo principal
#Ademas .size() se encarga de contar cuantos hay en total de cada, lo cual se almacena en una nueva columna
#que se llamara 'Total' gracias al reset del nombre de la columna de conteo
tipoPrinc = pokemon.groupby('Type 1').size().reset_index(name='Total')
#Por otro lado, .sort_values() se encarga de ordenar los valores segun la columna 'Total' de forma ascendente
tipoPrinc = tipoPrinc.sort_values(by='Total', ascending=True)
#Para terminar mostramos el nuevo dataframe
tipoPrinc
```
